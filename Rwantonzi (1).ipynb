```python
#  Writig a Python program that takes your full name as a string and: prints it in uppercase

full_name = "Rwantonzi Thomas"  # Replacig with your actual full name
uppercase_name = full_name.upper()

print(f"Original Name: {full_name}")
print(f"Uppercase Name: {uppercase_name}")
```

```python
# Creating a list of 5 different fruits. print the first and last fruit using indexing

fruits = ["apple", "banana", "cherry", "date", "elderberry"]

# The first fruit is at index 0
first_fruit = fruits[0]

# The last fruit is at index -1 (or len(fruits) - 1)
last_fruit = fruits[-1]

print(f"List of fruits: {fruits}")
print(f"First fruit: {first_fruit}")
print(f"Last fruit: {last_fruit}")
```

```python
# Creating a dictionary called student with keys: "name", "age", and "course". print student's name

student = {
    "name": "Thomas Rwantonzi",
    "age": 21,
    "course": "Computer Science"
}

student_name = student["name"]

print(f"Student Dictionary: {student}")
print(f"Student's Name: {student_name}")
```
